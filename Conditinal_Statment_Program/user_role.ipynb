47. Write a program to check user role (admin, user, guest) using match case.


```python
role = (input("Enter a Role ="))

match role :
    case 'Admin':
        print("Welcome Admin , You have full Access.")
    case 'User':
        print("Wlcome User , You have Normal Access.")
    case 'Guest':
        print("Welcome Guest , You hava Limited Access.")
    case _ :
        print("Invalid Role.")
```
