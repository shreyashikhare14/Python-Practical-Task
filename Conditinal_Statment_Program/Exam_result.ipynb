24. Write a program to check exam results (Distinction, First Class, Pass, Fail).


```python
marks = int(input("Enter a marks ="))
if marks > 100 or marks < 0:
    print("Invaild Marks")
elif marks >= 75:
    print("Your result is Distinction")
elif marks >=60:
    print("First Class ")
elif marks >=35:
    print("You are Pass")
else:
    print("you are fail ")
```
