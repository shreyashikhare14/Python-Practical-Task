26. Write a program to check month names based on month number.


```python
number = int(input("Enter a number:"))
if number == 1:
    print("Jan")
elif number == 2:
    print("Feb")
elif number == 3:
    print("March")
elif number == 4:
    print("April")
elif number == 5:
    print("May")
elif number == 6:
    print("June")
elif number == 7:
    print("July")
elif number == 8:
    print("Aug")
elif number == 9:
    print("Sep")
elif number ==10:
    print("Oct")
elif number== 11:
    print("Nov")
elif number == 12:
    print("Dec")
else:
    print("Invalid Month Number")
```
