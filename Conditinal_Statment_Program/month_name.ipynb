26. Write a program to check month names based on month number.


```python
month = int(input("Enter a Month Number ="))
match month:
    case 1 :
        print("Jan")
    case 2 :
        print("Feb")
    case 3 :
        print("March")
    case 4 :
        print("April")
    case 5 :
        print("May")
    case 6 :
        print("June")
    case 7:
        print("July")
    case 8:
        print("Aug")
    case 9:
        print("Sep")
    case 10:
        print("Acto")
    case 11:
        print("Nov")
    case 12:
        print("Dec")
    case _:
        print("Invalid Month")


        
    
```
