50. Write a program to build a simple ATM menu using a match case.


```python
menu = int(input("Enter a Menu = 1.Check Balance \n 2. Withdraw Cash \n 3.Deposit Money \n 4. Set Pin"))

match menu :
    case 1 :
        print("Check Balance.")
    case 2 :
        print("Withdraw Cash.")
    case 3 :
        print("Deposit Money.")
    case 4 :
        print("Set Pin.")
    case _:
        print("Invalid Menu.")
```
