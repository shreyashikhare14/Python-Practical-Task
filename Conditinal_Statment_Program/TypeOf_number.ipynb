30. Write a program to check the type of number (single, double, or three digit).

```python
digit = int(input("Enter a Digit ="))
digit = abs(digit)

if digit >=0 and digit <=9:
    print(digit,"is a Single Digit Number.")
elif digit >=10 and digit <=99:
    print(digit,"is a Double Number.")
elif digit >=100 and digit <= 999:
    print(digit,"is a Three digit")
else:
    print(digit,"is more than three digit") 
```
