46. Write a program to print grade description using match case.


```python
marks = int(input("Enter your marks ="))

match marks :
    case m if marks >=0 and  marks <35:
                print("You are Fail ")
    case m if marks >= 35 and marks <50:
            print("Grade C")
    case m if marks >=50 and marks <70 :
            print("Grade B")
    case  m if marks >=70 and marks <90:
            print("Grade A")
    case m if marks >=90 and marks <100:
            print("Grade A+")
    case _ :
        print("Invalid Marks")

```
