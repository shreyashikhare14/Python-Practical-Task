48. Write a program to print language choice using match case.


```python
language = input("Enter a Language =").capitalize()

match language :
    case 'English':
        print("Welcome , You Selected English Language.")
    case 'Hindi' :
        print("Welcome , You Selected Hindi Language. ")
    case 'Marathi':
        print("Welcome , You Selected Marathi Language.")
    case 'Japanese':
        print("Welcome , You Selected Japanese Language.")
    case _ :
        print("This Language is Not Avilable.")
```
