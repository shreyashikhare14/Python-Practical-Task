49. Write a program to print result status using match case.


```python
result = input("Enter a Result Status(Pass,Fail,ATKT) = ").capitalize()

match result :
    case 'Pass':
        print("Congratulaions,You are Pass.")
    case 'Fail':
        print("Sorry, You are Fail in the Exam.")
    case 'Atkt':
        print("You have a ATKT in the Exam.")
    case _ :
        print("Invalid Result Status.")
```
