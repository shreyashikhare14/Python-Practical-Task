38. Write a program to check whether a student passed all subjects.


```python
Science = int(input("Enter a Science Marks = "))
Maths = int(input("Enter a Maths Marks ="))
English = int(input("Enter a English Marks ="))

if Science >=35 and Maths >=35 and English >=35:
    print("You are Passed in all Subjects.")
else:
    print("You are Not Passed in all Subject.")
```
