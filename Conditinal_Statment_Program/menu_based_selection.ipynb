45. Write a program to print menu-based food selection using match case.


```python
number =int(input("Enter a Category (1,2,3,4) ="))

match number :
    case 1 :
        print("Starter Menu.")
    case 2:
        print("Veg Menu")
    case 3:
        print("Non-Veg Menu")
    case 4:
        print("Deserts Menu")
    case _:
        print("This Menu is Not Avilable.")
```
