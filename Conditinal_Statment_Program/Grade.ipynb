17. Write a program to check grade based on marks (A, B, C, Fail).

```python
marks = int(input("Enter your marks = "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks > 80:
    print("A Grade")
elif marks > 60:
    print("B Grade")
elif marks >= 35:
    print("C Grade")
else:
    print("You are Fail")
```
