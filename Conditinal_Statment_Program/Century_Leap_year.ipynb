34. Write a program to check whether a year is a leap year and century year.

```python
year = int(input("Enter a Year ="))

if year % 400 == 0:
    print(year , "is a Century Year and Leap Year")
elif year % 100 ==0:
    print(year,"is a Century Year and But Not a Leap Year.")
elif year % 4 == 0:
    print(year,"is a Century Year and But Not a Leap Year.")
else:
    print(year,"is Neither a Leap Year nor a Century Year.")
```
