22. Write a program to check salary bonus based on years of service.

```python
service_year = int(input("Enter a year of serive = "))

if service_year >=10:
    print("Salary Bonus is 10000 Rs")
elif service_year >= 5:
    print("Salary Bonus is 5000")
elif service_year >=0:
    print("No Bonus")
else:
    print("Invalid Service")
```
