14. Write a program to check whether a given character is uppercase or lowercase.

```python
char_name = input("Enter a character =")
if char_name >= 'A' and char_name <= 'Z' :
    print(char_name," character is Uppercase")
elif char_name >='a' and char_name <='z':
    print(char_name," character is Lowercase")
else:
    print("Not Applicable")
```
