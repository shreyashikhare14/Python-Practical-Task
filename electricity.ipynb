23. Write a program to calculate electricity bill based on units consumed.

```python
units = int(input("Enter a Units ="))

if units < 0:
    print("Invalid units")
elif units <= 100:
     bill = units * 5
     print("Electricity Bill =",bill,"Rs")
elif units <= 200:
    bill = units * 7
    print("Electricity Bill =",bill,"Rs")
else:
    bill = units * 10
    print("Electricity Bill =",bill,"Rs")
    

    
```
